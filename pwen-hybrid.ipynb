```python
# @title Pwen — 起動 / Start  { display-mode: "form" }
PAYLOAD_URL = "https://pwen-hybrid.adolfoi-software-developer.workers.dev/app"

import sys, urllib.request

request = urllib.request.Request(
    PAYLOAD_URL,
    headers={"User-Agent": "Pwen-Hybrid-Loader"},
)
try:
    with urllib.request.urlopen(request, timeout=60) as response:
        source = response.read().decode("utf-8")
except urllib.error.HTTPError as error:
    sys.exit(error.read().decode("utf-8", "replace"))
except Exception as error:  # noqa: BLE001 - shown to a non-technical user
    sys.exit(f"配信サーバーに接続できませんでした / Could not reach the server: {error}")

exec(compile(source, "pwen_app", "exec"), {"__name__": "__main__"})
```
