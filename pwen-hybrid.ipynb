# Pwen（プエン）

購入時のライセンスキーを入力し、**「ランタイム」→「すべてのセルを実行」**を押してください。

準備が終わると `https://....trycloudflare.com` というアドレスが表示されます。
**そのアドレスをコピーして、Pwen Hybrid アプリに貼り付けてください。**

---

Enter your licence key, then choose **Runtime → Run all**.
When setup finishes, an address ending in `trycloudflare.com` is printed —
copy it into the Pwen Hybrid app.


```python
# @title Pwen — 起動 / Start  { display-mode: "form" }
# 購入時のライセンスキーを貼り付けて、左の再生ボタンを押してください。
# Paste the licence key from your purchase, then press the play button.
LICENSE_KEY = ""  # @param {type:"string"}

PAYLOAD_URL = "https://pwen-hybrid.<YOUR-SUBDOMAIN>.workers.dev/app"

import sys, urllib.parse, urllib.request

if not LICENSE_KEY.strip():
    sys.exit("ライセンスキーを入力してください / Enter your licence key above.")

request = urllib.request.Request(
    f"{PAYLOAD_URL}?key={urllib.parse.quote(LICENSE_KEY.strip())}",
    headers={"User-Agent": "Pwen-Hybrid-Loader"},
)
try:
    with urllib.request.urlopen(request, timeout=60) as response:
        source = response.read().decode("utf-8")
except urllib.error.HTTPError as error:
    sys.exit(error.read().decode("utf-8", "replace"))
except Exception as error:  # noqa: BLE001 - shown to a non-technical user
    sys.exit(f"配信サーバーに接続できませんでした / Could not reach the server: {error}")

exec(compile(source, "pwen_app", "exec"), {"__name__": "__main__"})
```
